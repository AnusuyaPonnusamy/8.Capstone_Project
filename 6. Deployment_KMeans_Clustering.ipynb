```python
import pickle

print("Please enter the input values in the following order:")

print("1. discounted_price")
print("2. discount_percentage")
print("3. rating")
print("4. rating_count")

print("\nExample:")
print("[499, 40, 4.3, 5000]")


loaded_model = pickle.load(
    open("finalized_model_KMeans.sav", "rb")
)


kmeans = loaded_model["kmeans"]

scaler = loaded_model["scaler"]

cluster_mapping = loaded_model["cluster_mapping"]

rating_count_max = loaded_model["rating_count_max"]

discounted_price_max = loaded_model["discounted_price_max"]


print("Please enter the following product details:")

discounted_price = float(
    input("Enter discounted price: ")
)

discount_percentage = float(
    input("Enter discount percentage: ")
)

rating = float(
    input("Enter rating: ")
)

rating_count = float(
    input("Enter rating count: ")
)

rating_score = (rating / 5) * 100

demand_score = (
    rating_count / rating_count_max
) * 100

discount_score = discount_percentage

price_score = (
    1 - discounted_price / discounted_price_max
) * 100


Opportunity_Score = (
    0.30 * rating_score +
    0.30 * demand_score +
    0.20 * discount_score +
    0.20 * price_score
)


X_new = [[
    discounted_price,
    discount_percentage,
    rating,
    rating_count,
    Opportunity_Score
]]


X_new_scaled = scaler.transform(X_new)


cluster = kmeans.predict(X_new_scaled)[0]


result = cluster_mapping[cluster]


print("\nFinal Recommendation:")
print(result)
```
