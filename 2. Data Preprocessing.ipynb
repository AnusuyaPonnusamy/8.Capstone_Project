```python
dataset = dataset.drop_duplicates()
```

```python
# Step 6 : Remove ₹ Symbol
```

```python
dataset["discounted_price"] = dataset["discounted_price"].str.replace("₹","",regex=False)
```

```python
dataset["actual_price"] = dataset["actual_price"].str.replace("₹","",regex=False)
```

```python
dataset["discounted_price"], dataset["actual_price"]
```

```python
# Step 7 : Remove Commas
```

```python
dataset["discounted_price"] = dataset["discounted_price"].str.replace(",","",regex=False)
```

```python
dataset["actual_price"] = dataset["actual_price"].str.replace(",","",regex=False)
```

```python
dataset["rating_count"] = dataset["rating_count"].str.replace(",","",regex=False)
```

```python
dataset["discounted_price"], dataset["actual_price"], dataset["rating_count"]
```

```python
# Step 8 : Remove Percentage Symbol
```

```python
dataset["discount_percentage"] = dataset["discount_percentage"].str.replace("%","",regex=False)
```

```python
dataset["discount_percentage"] 
```

```python
# Step 9 : Remove | Symbol
dataset["rating"] = dataset["rating"].str.replace("|","",regex=False)
```

```python
# Step 10 : Convert Numeric Columns
```

```python
dataset.dtypes
```

```python
dataset["discounted_price"] = pd.to_numeric(dataset["discounted_price"])
```

```python
dataset["actual_price"] = pd.to_numeric(dataset["actual_price"])
```

```python
dataset["discount_percentage"] = pd.to_numeric(dataset["discount_percentage"])
```

```python
dataset["rating"] = pd.to_numeric(dataset["rating"])
```

```python
dataset["rating_count"] = pd.to_numeric(dataset["rating_count"])
```

```python
dataset.dtypes
```

```python
# Step 11 : Handle Missing Values
```

```python
dataset.isnull().sum()
```

```python
dataset["rating_count"] = dataset["rating_count"].fillna(dataset["rating_count"].median())
```

```python
dataset["rating"] = dataset["rating"].fillna(dataset["rating"].median())
```

```python
dataset.isnull().sum()
```

```python
# Step 12 : Remove Leading and Trailing Spaces
```

```python
cols=dataset.select_dtypes(include="object").columns
```

```python
for column in cols:
    dataset[column] = dataset[column].str.strip()
```

```python
# Step 13 : Convert Text to Lowercase
```

```python
dataset["product_name"] = dataset["product_name"].str.lower()

dataset["category"] = dataset["category"].str.lower()

dataset["about_product"] = dataset["about_product"].str.lower()

dataset["review_title"] = dataset["review_title"].str.lower()

dataset["review_content"] = dataset["review_content"].str.lower()
```

```python
# Step 14 : Remove Special Characters
```

```python
dataset["product_name"] = dataset["product_name"].str.replace(r"[^a-zA-Z0-9 ]","",regex=True)

dataset["about_product"] = dataset["about_product"].str.replace(r"[^a-zA-Z0-9 ]","",regex=True)

dataset["review_title"] = dataset["review_title"].str.replace(r"[^a-zA-Z0-9 ]","",regex=True)

dataset["review_content"] = dataset["review_content"].str.replace(r"[^a-zA-Z0-9 ]","",regex=True)
```

```python
# Outliers Analysis
```

Boxplots were generated for all numerical variables to identify extreme values (potential outliers).

Discounted Price
----------------
Observation: 
- Several products have prices much higher than the rest of the products.
- Most products are concentrated in the lower price range, while a few premium products appear as outliers.

Interpretation:
- This dataset contains products ranging from inexpensive accessories (e.g., cables, chargers, earphones) to premium electronics. High-priced products naturally appear as outliers.
- These are not data errors. They represent premium products with higher selling prices.

Decision: Retain these outliers.

Actual Price
------------
Observation: 
- Similar to discounted price, there are several high-value outliers.

Interpretation: 
- Products such as premium headphones, smartwatches, or branded accessories have significantly higher MRP than regular accessories.
- These products may generate higher profit margins and should remain in the dataset.

Decision: Retain the outliers.

Discount Percentage
-------------------
Observation: 
- Some products have extremely high discounts.

Interpretation: 
- Large discounts are common on Amazon due to flash sales, festive offers, and promotional campaigns.
- High discount values are valid business observations rather than incorrect data.

Decision: Retain these records.

Rating
------ 
Observation:
- Very few or no significant outliers were observed.
- Ratings are bounded between 1 and 5.

Interpretation: 
- The ratings fall within the expected range for Amazon products.
- The rating values appear consistent.

Decision: Retain these outliers.

Rating Count
------------
Observation: 
- A large number of products have extremely high review counts compared with the majority.

Interpretation:
- Popular products naturally accumulate thousands of customer reviews while newer products receive very few.
- High review counts indicate high customer demand and product popularity. These observations are valuable for predicting demand.

Decision: Do not remove these outliers.

---------------------------------------------------------------------------------------------------------------------------

Overall Interpretation:
-----------------------
- Although boxplots indicate the presence of outliers in several numerical variables, these observations correspond to genuine business scenarios rather than data quality issues.
- For an Amazon marketplace dataset:
   - Premium products naturally have higher prices.
   - Best-selling products naturally have significantly higher review counts.
   - Promotional campaigns often lead to unusually high discounts.
- Therefore, these outliers provide important business information and should be retained for further analysis instead of being removed.

---------------------------------------------------------------------------------------------------------------------------
Conclusion:
-----------
Outlier analysis was performed using boxplots. The detected outliers represent legitimate business observations (premium-priced products, highly reviewed products, and products with high promotional discounts). Not using the IQR method to remove outliers in this dataset. Removing them could eliminate valuable information about premium or highly popular products, which are precisely the kinds of products an Amazon seller may want to identify. Since these values are meaningful and not data entry errors, no outlier removal was performed. The dataset was retained in its original form for subsequent exploratory analysis and model development.

numeric_columns = dataset.select_dtypes(include=np.number).columns

for column in numeric_columns:

    Q3 = dataset[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR

    upper = Q3 + 1.5*IQR

    dataset = dataset[(dataset[column] >= lower) &
                      (dataset[column] <= upper)]
