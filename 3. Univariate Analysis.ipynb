# UNIVARIATE ANALYSIS

```python
# Import Libraries
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dataset = pd.read_csv("Clean_Amazon_Dataset.csv")
```

```python
# CONTINUOUS VARIABLES
```

```python
continuous_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]
```

# Measure of Location of the Data

```python
# 1. Frequency Distribution - Shows how many times each value occurs in the dataset.
```

```python
for column in continuous_columns:
    print(dataset[column].value_counts())
```

```python
# 2. Cumulative Frequency - Shows the running total of observations as values increase.
```

```python
for column in continuous_columns:
    print(dataset[column].value_counts().sort_index().cumsum())
```

```python
# 3. Percentiles - Divide the data into 100 equal parts to understand the distribution of values.
```

```python
for column in continuous_columns:
    print(dataset[column].quantile([0.25,0.50,0.75,0.90,0.95]))
```

```python
# 4. Interquartile Range (IQR) - Measures the spread of the middle 50% of the data
```

```python
for column in continuous_columns:

    Q1 = dataset[column].quantile(0.25)

    Q3 = dataset[column].quantile(0.75)

    IQR = Q3 - Q1

    print(column)

    print("Q1 :",Q1)

    print("Q3 :",Q3)

    print("IQR :",IQR)
```

```python
# 5. Range - Difference between the maximum and minimum values.
```

```python
for column in continuous_columns:

    print(column)

    print("Minimum :",dataset[column].min())

    print("Maximum :",dataset[column].max())

    print("Range :",dataset[column].max()-dataset[column].min())
```

# Measure of Central Tendency

```python
# 6. Mean - The arithmetic average of all observations
```

```python
for column in continuous_columns:

    print(column)

    print("Mean :",dataset[column].mean())
```

```python
# Mean Interpretation: The average product rating is above 4, indicating generally positive customer satisfaction.                          |
```

```python
# 7. Median - The middle value after sorting the data. 
```

```python
for column in continuous_columns:

    print(column)

    print("Median :",dataset[column].median())
```

```python
# 8. Mode - The value that occurs most frequently.  
```

```python
for column in continuous_columns:

    print(column)

    print("Mode :")

    print(dataset[column].mode())
```

# Measure of Spread

```python
# 9. Variance - Measures how much the values differ from the mean
```

```python
for column in continuous_columns:

    print(column)

    print("Variance :",dataset[column].var())
```

```python
# 10. Standard Deviation - Measures the average spread of values around the mean.      
```

```python
for column in continuous_columns:

    print(column)

    print("Standard Deviation :",dataset[column].std())
```

```python
# 11. Z-Score
```

```python
for column in continuous_columns:

    dataset[column+"_zscore"] = (
        dataset[column]-dataset[column].mean()
    )/dataset[column].std()

dataset.head()
```

# Histogram and Skewness

```python
# 12. Histogram - Displays the distribution of numerical data using bars.
```

```python
for column in continuous_columns:

    plt.figure(figsize=(7,4))

    plt.hist(dataset[column],bins=30)

    plt.title(column)

    plt.xlabel(column)

    plt.ylabel("Frequency")

    plt.show()
```

```python
# 13. Skewness - Measures whether the distribution is symmetric or skewed.
```

```python
for column in continuous_columns:

    print(column)

    print("Skewness :",dataset[column].skew())
```

```python
# 14. Kurtosis - Measures how peaked or heavy-tailed a distribution is. 
```

```python
for column in continuous_columns:

    print(column)

    print("Kurtosis :",dataset[column].kurt())
```

# ADVANCED STATISTICS ANALYSIS

```python
# 15. Normal Distribution - A bell-shaped distribution where data is symmetrically distributed around the mean. 
```

```python
from scipy.stats import norm
import matplotlib.pyplot as plt

for column in continuous_columns:

    plt.figure(figsize=(7,4))

    mean = dataset[column].mean()
    std = dataset[column].std()

    x = dataset[column]

    plt.hist(x, bins=30, density=True)

    xmin, xmax = plt.xlim()

    x_axis = np.linspace(xmin, xmax, 100)

    plt.plot(x_axis, norm.pdf(x_axis, mean, std))

    plt.title("Normal Distribution - " + column)

    plt.show()
```

```python
import seaborn as sns
for column in continuous_columns:
    sns.distplot(dataset[column])
```

```python
# 16. Probability Density Function (PDF)
```

```python
from scipy.stats import norm

for column in continuous_columns:

    mean = dataset[column].mean()
    std = dataset[column].std()

    print("\n", column)

    value = mean

    probability = norm.pdf(value, mean, std)

    print("Mean :", mean)

    print("Standard Deviation :", std)

    print("PDF at Mean :", probability)
```

```python
# 17. Cumulative Distribution Function (CDF)
```

```python
from scipy.stats import norm

for column in continuous_columns:

    mean = dataset[column].mean()

    std = dataset[column].std()

    value = mean

    cumulative_probability = norm.cdf(value, mean, std)

    print(column)

    print("CDF :", cumulative_probability)
```

```python
# 18. Standard Normal Distribution (Z-Distribution)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

continuous_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

dataset_std = pd.DataFrame(
    scaler.fit_transform(dataset[continuous_columns]),
    columns=continuous_columns
)

dataset_std.head()
```

```python
# Plot Standard Normal Distribution
```

```python
for column in dataset_std.columns:

    plt.figure(figsize=(7,4))

    plt.hist(dataset_std[column], bins=30)

    plt.title("Standard Normal Distribution - " + column)

    plt.show()
```
